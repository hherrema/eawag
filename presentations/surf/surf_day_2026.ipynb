```python
# imports
import ewgreaders as ewg
import matplotlib.pyplot as plt
import seaborn as sns
import time
```

```python
def reveal_all_secrets_of_the_lake():
    lake = 'Zug'
    di = ewg.get_data_index(lake, 'ctd')

    pl = 'P10'
    di_sel = di.query("profile_loc == @pl")

    p = di_sel.iloc[-2]
    ctdreader = ewg.ctd_reader.CTDReader(p.lake, p.date, p.fname)
    ds = ctdreader.load()

    time.sleep(5)

    fig, ax = plt.subplots(figsize=(11, 7.5))

    sns.lineplot(x=ds['temp'], y=ds['depth'], orient='y', color='tab:blue', lw=2, ax=ax)

    ax.set_xlabel('Temperature [°]', fontsize=14, labelpad=10)
    ax.set_ylabel('Depth [m]', fontsize=14, labelpad=10)
    ax.tick_params(axis='both', labelsize=12)

    ax.grid(True)
    ax.invert_yaxis()

    plt.show()


def reveal_more_secrets_of_the_lake():
    lake = 'Zug'
    di = ewg.get_data_index(lake, 'ctd')

    pl = 'P10'
    di_sel = di.query("profile_loc == @pl")

    p = di_sel.iloc[-2]
    ctdreader = ewg.ctd_reader.CTDReader(p.lake, p.date, p.fname)
    ds = ctdreader.load()

    temp_max = ds['temp'].max().item()
    temp_min = ds['temp'].min().item()

    if temp_max < 32:
        print(f'The maximum temperature in the lake is {temp_max} degrees.  IT IS FROZEN!')

    if temp_min < 32:
        print(f'The minimum temperature in the lake is {temp_min} degrees.  ICE GLACE EIS!')


def add_zero_to_number(x):
    """
    Add zero to a number.

    Parameters
    ----------
    x : float
        Number to add zero to.

    Returns
    -------
    new_x : float
        Original number plus zero.
    """
    new_x = x + 0

    return new_x


def multiply_by_1(x):
    """
    Multiply a number by one.

    Parameters
    ----------
    x : float
        Number to multiply by one.
    
    Returns
    -------
    new_x : float
        Original number times one.
    """
    # 0 times anything equals 0
    if x == 0:
        return 0
    
    new_x = x * 1

    return new_x
    
```
